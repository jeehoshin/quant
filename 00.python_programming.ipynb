### 2.7.1 클래스

데이터 멤버인 속성과 동작을 수행하는 메서드로 구성됨.  
클래스로부터 객체를 생성하는 것을 인스턴스화 라고 하며, 생성된 인스턴스가 가지고 있는 속성과 메서드는 . 표기법을 사용해서 접근

````python
class my_first_class : 
    
    #클래스 변수를 선언
    cls_var = "가나다라"
    
    #인스턴스 변수는 선언되지 않음
    def cls_method(self) : 
        print(my_first_class.cls_var + "\t dddd - ")
        
#클래스를 인스턴스화! 붕어빵 하나를 만들었다.         
mfc = my_first_class()

#클래스 변수에 접근한다.
print(mfc.cls_var)


#클래스 메서드를 호출한다.
mfc.cls_method()
````

####  클래스 상속

````python
class aa : 
    def method_a(self) : 
        print('메서드a')
    def method(self) : 
        print("메서드")
        
#우선 상속 받는 순간 부모 클래스의 모든 메서드를 사용할 수 있다.
class bb(aa) : 
    
    #이렇게 부모 메서드를 상속받아서 다른 이름으로 사용할 수 있다.
    def method_a2(self) : 
        super().method_a()
        
    #자식 클래스에서만 사용할 수 있는 메서드를 구축할 수 있다.
    def method_b(self) : 
        print('메서드b')
        


class cc(aa) : 
    #인스턴스 변수를 선언!
    def __init__(self) : 
        super()
    
    def method_c(self) : 
        print('메서드c')
        
````

````python
instance1 = bb()
instance2 = cc()
````

### 2.7.4 클래스 메서드

#### __init__ 생성자

인스턴스가 생성될 때 자동으로 호출되는 메서드! 이건 인스턴스 내부로 국한되게 된다.

````python
class nasdaq_stock : 
    '''나스닥 주식에 대한 class''' #독스트링
    
    #클래스 변수 선언
    count = 0 
    symbol = "NASDAQS"
    
    #__init__은 인스턴스가 생성될 때 자동으로 호출되는 함수
    def __init__(self, symbol, price) : 
        """나스닥의 constructor"""
        #이 인스턴스가 만들어질때 자동으로 객체가 선언됨
        self.symbol = symbol
        self.price = price
        
        #클래스변수 count 더해줌 
        nasdaq_stock.count += 1 
        
        print("calling __init__({0},{1:.2f}) > count : {2}".format(self.symbol, self.price, nasdaq_stock.count))
        
    def test_func(self) : 
        """해당 종목의 가격을 알려줌 """
        print("이 종목의 티커는 {0} 이고, 가격은 {1:.2f}입니다".format(self.symbol, self.price))
        
    # __del__은 인스턴스가 삭제될 때 자동으로 호출되는 함수
    def __del__(self) : 
        '''없애버리자'''
        print('calling __del__{0}'.format(self))
````

````python
google = nasdaq_stock('GOOG', 1154.05)
````

````python
google.test_func()
````

````python
#해당 인스턴스를 지울때 메세지가 나온다
del(google)
````

````python
ms = nasdaq_stock('MSFT', 102.44)
````
