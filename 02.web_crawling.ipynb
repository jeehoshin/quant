### 4.1.2 read_html() 함수로 read

```python
import pandas as pd

krx_list = pd.read_html("C:/python/0500.quant/02.data/상장법인목록.xls")
krx_list
```

> 조금은 이상하게 나온다...

```python
"""제대로 해서 뽑으려면 인덱스에 [0]을 넣어주면 된다"""
krx_list[0]
```

- map 함수로 보정  
위에서 볼 수 있듯, 종목코드가 앞 0이 제외되어 뽑힌다

```python
krx_list[0].종목코드 = krx_list[0].종목코드.map("{:06d}".format)
krx_list[0]
```

> 우선 map 함수는 시리즈에서 인수로 받은 함수로 값을 변경한다고 한다.  
`map(변환 함수, 순회 가능한 데이터)`
  
> {:06d}의 의미는 6자리 숫자 형식으로 뽑되, 앞을 0으로 채우라는 의미라고 한다. 

- url로 읽어들이기  
url로 바로 읽어들일 수 있다고 한다!

```python
"""
대신 krx 홈페이지에서 엑셀 다운로드의 경로를 추출해줘야 한다.
"""
krx_list2 = pd.read_html("https://kind.krx.co.kr/corpgeneral/corpList.do?method=download&pageIndex=1&currentPageSize=3000&comAbbrv=&beginIndex=&orderMode=3&orderStat=D&isurCd=&repIsuSrtCd=&searchCodeType=&marketType=&searchType=13&industry=&fiscalYearEnd=all&comAbbrvTmp=&location=all")[0]
```

```python
krx_list2['종목코드'] = krx_list2['종목코드'].map('{:06d}'.format)
krx_list2
```

```python
krx_list2 = krx_list2.sort_values(by = '종목코드')
krx_list2
```

### 4.4.3 맨 뒤 페이지 숫자 구하기 

```python
from bs4 import BeautifulSoup
from urllib.request import urlopen
import requests

#우선 분석하고자 하는 url을 긁어옴
url = "https://finance.naver.com/item/sise_day.naver?code=000720&page=2"

with urlopen(url) as doc : 
    html = BeautifulSoup(doc, 'lxml')
    pgrr = html.find('td', class_= 'pdRR')
    print(pgrr.a['href'])
```
